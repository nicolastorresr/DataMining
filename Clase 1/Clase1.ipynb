# Distancias

<div align="right">
    <img src="img/euc-manh.png" alt="" width="40%" align="right" />
</div>

La **distancia Minkowski** de orden $p$ entre dos puntos $ \mathbf{x} = (x_1, x_2, \dots , x_n )$, $\mathbf{y} = (y_1, y_2, \dots , y_n) \in \mathbb{R}^n $, se define como:

$$ D(\mathbf{x},\mathbf{y}) = \left( \sum_{i=1}^n |x_i - y_i|^p \right) ^ {1/p} $$

# Similaridad

<div align="right">
    <img src="img/euc-cos.png" alt="" width="40%" align="right" />
</div>


La **similitud coseno** mide el coseno del ángulo entre dos vectores distintos de cero:

$$ \cos{(\theta)} = \frac{\mathbf{x} \cdot \mathbf{y}}{\|\mathbf{x}\| \|\mathbf{y}\|} = \frac{\sum\limits_{i=1}^n x_i y_i }{\sqrt{\sum\limits_{i=1}^n x_i^2} \sqrt{\sum\limits_{i=1}^n y_i^2} }  $$
