# ¿Qué es un modelo de lenguaje?

Un modelo de lenguaje es básicamente una distribución de probabilidad sobre palabras o secuencias de palabras. Entre otras cosas, pueden sugerir la siguiente palabra que escribimos.

## Probabilistic Language Model

Dada una secuencia de palabras $D$ en una oración, la tarea es calcular las probabilidades de todas las palabras que terminarían esta oración. Las palabras se eligen de un vocabulario dado ($V$). Es decir, se necesita calcular la siguiente probabilidad condicional:

$$ P\left( w_{D+1}^{(i)} | (w_{1}^{(i)},w_{2}^{(i)},\dots,w_{D}^{(i)}) \right) $$

para cualquier ejemplo dado ($i$). Cada palabra está representada por un vector $w$, que tiene la dimensión del tamaño del diccionario ($|V|$). Por ejemplo, considere: "I am happy because I am learning".

$$P(\text{happy}|\text{I am}) = \frac{C(\text{I am happy})}{C(\text{I am})} = \frac{1}{2} = 0.5$$

# Redes LSTM y GRU

- Debido al problema de las gradientes desvanecientes, las RNNs sufren de memoria a corto plazo y pueden "olvidar" las primeras palabras de la secuencia. 

- Para mitigar la memoria a corto plazo, se crearon dos redes neuronales recurrentes especializadas: Long Short-Term Memory (LSTM) y Gated Recurrent Units (GRU).

- LSTM y GRU son capaces de aprender secuencias de términos a largo plazo utilizando mecanismos llamados "*gates*". Estas "puertas" son diferentes operaciones tensoriales que pueden aprender qué información mantener y cual eliminar en la capa oculta.

# LSTM

- Las redes *Long Short Term Memory* (memoria a largo plazo), generalmente llamadas simplemente "LSTM", son un tipo especial de RNN, capaces de aprender dependencias a largo plazo.

- Cada unidad mantiene un estado de memoria de la celda $c_t$ en el tiempo $t$. 

- Tiene 3 compuertas: compuerta de olvido, compuerta de entrada y compuerta de salida.

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
```

```python
with open("data/quijote.txt", encoding="utf8") as f:
    text = f.read()
```

```python
print(text[:500])
```

```python
"".join(sorted(set(text.lower())))
```

```python
tokenizer = keras.preprocessing.text.Tokenizer(char_level=True)
tokenizer.fit_on_texts(text)
```

```python
max_id = len(tokenizer.word_index) # numero de caracteres distintos
max_id
```

```python
dataset_size = tokenizer.document_count # numero total de caracteres
dataset_size
```

```python
[encoded] = np.array(tokenizer.texts_to_sequences([text])) - 1
train_size = dataset_size * 90 // 100
dataset = tf.data.Dataset.from_tensor_slices(encoded[:train_size])
```

```python
n_steps = 100
window_length = n_steps + 1 # target = input shifted 1 character ahead
dataset = dataset.window(window_length, shift=1, drop_remainder=True)
```

```python
dataset = dataset.flat_map(lambda window: window.batch(window_length))
```

```python
batch_size = 32
dataset = dataset.shuffle(10000).batch(batch_size)
dataset = dataset.map(lambda windows: (windows[:, :-1], windows[:, 1:]))
```

```python
dataset = dataset.map(
    lambda X_batch, Y_batch: (tf.one_hot(X_batch, depth=max_id), Y_batch))
```

```python
dataset = dataset.prefetch(1)
```

```python
for X_batch, Y_batch in dataset.take(1):
    print(X_batch.shape, Y_batch.shape)
```

# Configurar el modelo

```python
model = keras.models.Sequential([
    keras.layers.GRU(128, return_sequences=True, input_shape=[None, max_id],
                     dropout=0.2),
    keras.layers.GRU(128, return_sequences=True,
                     dropout=0.2),
    keras.layers.TimeDistributed(keras.layers.Dense(max_id,
                                                    activation="softmax"))
])
```

```python
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam")
```

# Entrenar el modelo

Con el Quijote se demora varias horas en entrenar. Puede cargar los pesos y evitar el entrenamiento o bien usar menos datos para entrenar.

```python
model.load_weights("weights/cp.ckpt")
```

```python
checkpointer = tf.keras.callbacks.ModelCheckpoint(filepath='weights/cp.ckpt',
                               save_weights_only = True, verbose = 0)
```

```python
history = model.fit(dataset, epochs=10, callbacks=[checkpointer])
```

# Usar el modelo para generar texto

```python
def preprocess(texts):
    X = np.array(tokenizer.texts_to_sequences(texts)) - 1
    return tf.one_hot(X, max_id)
```

```python
def next_char(text):
    X_new = preprocess([text])
    y_proba = model(X_new)[0, -1:, :]
    rescaled_logits = tf.math.log(y_proba)
    char_id = tf.random.categorical(rescaled_logits, num_samples=1) + 1
    return tokenizer.sequences_to_texts(char_id.numpy())[0]
```

```python
next_char("don qui")
```

```python
def complete_text(text, n_chars=1):
    for _ in range(n_chars):
        text += next_char(text)
    return text
```
