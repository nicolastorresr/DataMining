### Capa Recurrente

Una capa **Recurrente** (`SimpleRNN`) es una Fully-connected RNN donde la salida en tiempo $t$, se usa para alimentar la entrada en tiempo $t+1$.

# El problema de las dependencias a largo plazo

Las RNNs sufren de memoria a corto plazo y van olvidando las primeras entradas de la secuencia. 

<pre>
... and that's why now I speak fluent _____. 
</pre>

Para mitigar este problema, se crearon dos redes neuronales recurrentes especializadas: Long Short-Term Memory (LSTM) y Gated Recurrent Units (GRU). Ambas son capaces de aprender secuencias a largo plazo, utilizando compuertas lógicas que realizan diferentes operaciones tensoriales, permitiendo aprender qué información mantener y qué información eliminar en la capa oculta.

### Long Short-Term Memory

En una LSTM, cada neurona mantiene un estado de memoria de la celda $c_t$ en el tiempo $t$ y tiene tres compuertas: de olvido, de entrada y de salida.

# Variaciones de BERT 

Existen muchas variaciones del modelo BERT. Un Lite BERT (ALBERT), por ejemplo, es una versión liviana de BERT donde la cantidad de parámetros (x18 menos parámetros) y el tiempo de entrenamiento (x1.7 más rápido) se reducen. El enfoque BERT robustamente optimizado (RoBERTa) es otra variación en la que las mejoras se realizan esencialmente entrenando BERT en un conjunto de datos más grande con lotes más grandes. Existen otras variaciones para diferentes idiomas, como BETO en español.

## Problema con seq2seq learning

La representación intermedia $\mathbf{z}$ no puede codificar información de todos los instantes de tiempo en la secuencia de entrada. Esto se conoce comúnmente como el problema del cuello de botella.

En teoría, las matemáticas indican que es posible capturar toda la información sobre la secuencia de entrada. Sin embargo, en la práctica, hasta dónde podemos ver en el pasado (la llamada ventana de referencia) es finita. Por lo tanto, las RNN tienden a olvidar información de los instantes de tiempo que están muy atrás.

# Atención al rescate!

<div align="center">
    <img src="img/attention.png" alt="" width="30%" align="center" />
</div>

La idea central es que el vector de contexto $\mathbf{z}$ debería tener acceso a todas las partes de la secuencia de entrada en lugar de solo a la última. En otras palabras, necesitamos formar una conexión directa con cada instante de tiempo.


Esto es lo que ahora llamamos atención, que es simplemente una noción de memoria, obtenida al prestar atención a múltiples entradas a lo largo del tiempo.

## Atención explícita

La red pondera su sensibilidad a la entrada en función de la memoria de entradas anteriores.

```python
GPT2 = pipeline('text-generation', model='gpt2')
```

```python
gen_text = GPT2('A golden retriever is a', num_return_sequences=5, 
                     max_length=100, pad_token_id=GPT2.tokenizer.eos_token_id)
```

```python
for k in gen_text:
    print(k['generated_text'].replace('\n',''),"\n")
```

```python
gen_text = GPT2('A golden retriever is not a', num_return_sequences=5, 
                     max_length=100, pad_token_id=GPT2.tokenizer.eos_token_id)
```

```python
for k in gen_text:
    print(k['generated_text'].replace('\n',''),"\n")
```
