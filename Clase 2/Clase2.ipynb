```python
def color_positive_green(val, threshold):
    """
    Takes a scalar and returns a string with
    the css property `'color: green'` for positive
    strings, black otherwise.
    """
    if isinstance(val, (int, float)) and val > threshold:
        color = 'green'
    else:
        color = 'black'
    return 'color: %s' % color
```

```python
df.style.applymap(color_positive_green, threshold=4)
```

```python
def color_positive_background(val, threshold):
    if isinstance(val, (int, float)) and val > threshold:
        background_color = 'background-color: lightgreen'
    else:
        background_color = 'background-color: white'
    return background_color
```

```python
df.style.applymap(color_positive_background, threshold=4)
```

# Dato faltante (Missing data)

```python
data_with_missing_values = {
    'Food': ['Chinese', None, 'Italian', 'Burgers', 'Chinese', 'Chinese', 
             'Burgers', 'Chinese', 'Italian'],
    'Age': [51, None, 82, 23, 46, None, None, 38, 31],
    'Distance': ['Close', None, None, 'Far', None, None, 'Very close', 'Close', 'Far'],
    'Company': ['Good', 'Good', 'Good', 'Bad', 'Good', 'Bad', 'Good', 'Bad', 'Good']
}

df_with_missing_values = pd.DataFrame(data_with_missing_values)
```

```python
df_with_missing_values.style.format(na_rep="Missing").highlight_null(color='red')
```

# Manejando valores faltantes

## Imputación de media/mediana: 

Para columnas numéricas como "Edad", puede reemplazar los valores faltantes con la media o mediana de los valores existentes en esa columna.

```python
def color_value(val, value):
    if pd.isna(val) or val is None:
        background_color = 'background-color: red'
    elif val == value:
        background_color = 'background-color: lightgreen'
    else:
        background_color = 'background-color: white'
    return background_color
```

```python
df_with_missing_values['Age'].fillna(df_with_missing_values['Age'].mean(), inplace=True)
x_age = df_with_missing_values['Age'].mean()
df_with_missing_values.style.applymap(lambda x: color_value(x, x_age))
```

## Imputación por moda

Para columnas categóricas como "Comida" y "Distancia", puede reemplazar los valores faltantes con la moda (valor más frecuente) de los valores existentes en esa columna.

```python
df_with_missing_values['Food'].fillna(df_with_missing_values['Food'].mode()[0], inplace=True)
df_with_missing_values['Distance'].fillna(df_with_missing_values['Distance'].mode()[0], inplace=True)
df_with_missing_values
```

```python
data = {
    'Food': ['Chinese', 'Italian', 'Italian', 'Burgers', 
             'Chinese', 'Chinese', 'Burgers', 'Chinese', 'Italian'],
    'Age': [51, 43, 82, 23, 46, 29, 42, 38, 31],
    'Distance': ['Close', 'Very close', 'Close', 'Far', 
                 'Very far', 'Too far', 'Very far', 'Close', 'Far'],
    'Company': ['Good', 'Good', 'Good', 'Bad', 'Good', 'Bad', 'Good', 'Bad', 'Good']
}

df_food = pd.DataFrame(data)
```

```python
df_food
```

```python
# Perform one-hot encoding on the "Food" column without prefix
df_onehot = pd.get_dummies(df_food['Food'])

# Concatenate the one-hot encoded columns with the original DataFrame
df_encoding = pd.concat([df_food, df_onehot], axis=1)

# Drop the original "Food" column
df_encoding = df_encoding.drop('Food', axis=1)

# Mapping for 'Distance' and 'Company'
distance_mapping = {'Close': 0, 'Very close': 1, 'Far': 2, 'Very far': 3, 'Too far': 4}
company_mapping = {'Good': 1, 'Bad': -1}

# Apply mapping to 'Distance' and 'Company' columns
df_encoding['Distance'] = df_encoding['Distance'].map(distance_mapping).fillna(0).astype(int)
df_encoding['Company'] = df_encoding['Company'].map(company_mapping).fillna(0).astype(int)

column_order = [3, 4, 5, 0, 1, 2]  # Change the order based on your requirements

# Reorder columns based on the specified order
df_encoding = df_encoding[df_encoding.columns[column_order]]
```

```python
df_encoding
```

```python
scaler = StandardScaler()
df_std = scaler.fit_transform(df_encoding)
df_std = pd.DataFrame(data = df_std,columns = df_encoding.columns)
```
