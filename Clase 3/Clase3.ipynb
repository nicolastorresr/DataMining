## KDE

Un gráfico de estimación de densidad del núcleo (KDE) es un método para visualizar la distribución de observaciones en un conjunto de datos, análogo a un histograma, pero en lugar de usar intervalos discretos, utiliza una función continua (kernel) para suavizar la distribución de los datos.

```python
plt.figure(figsize=(7,3))
sns.kdeplot(data=titanic_df['Age'], fill=True, common_norm=False, bw_method=2)
plt.xlabel('Age');plt.ylabel('Density');plt.title('Kernel Density Estimate of Age Distribution on Titanic')
plt.xlim(0,80);plt.show()
```

## Box plot

Representa gráficamente una serie de datos numéricos a través de la mediana (Q2), los cuartiles superior (Q3) e inferior (Q1), y los valores atípicos (outliers) que están fuera del rango intercuartílico (IQR = Q3 - Q1). Lower = Q1 - 1.5 * IQR y Upper = Q3 + 1.5 * IQR.

```python
explain_boxplot()
```

```python
# Generate example data with more measurements for each month
data = []

# Generate example data
months = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 
          'September', 'October', 'November', 'December']

# Classify months as 'hot' or 'cold'
hot_months = ['June', 'July', 'August']
cold_months = ['December', 'January', 'February']

# Generate temperatures based on seasons
mean_temperatures = [np.random.uniform(70, 90) if month in hot_months else np.random.uniform(20, 50) for month in months]

for month in months:
    #temperatures = np.random.uniform(20, 90, 100)  # Generate 100 temperature measurements for each month
    temperatures = [np.random.uniform(70, 90) if month in hot_months else np.random.uniform(20, 50) for _ in range(100)]    
    data.extend(zip([month[:3]] * len(temperatures), temperatures))

temperature_df = pd.DataFrame(data, columns=['Month', 'Temperature'])
```

# Treemap

```python
df = px.data.gapminder().query("year == 2007")
fig = px.treemap(df, path=[px.Constant("world"), 'continent', 'country'], values='pop',
                  color='lifeExp', hover_data=['iso_alpha'], color_continuous_scale='RdBu',
                  color_continuous_midpoint=np.average(df['lifeExp'], weights=df['pop']))
fig.update_layout(margin = dict(t=50, l=25, r=25, b=25))
fig.show()
```

# Nube de palabras

Representación visual de las palabras que conforman un texto de modo que las más frecuentes se muestran con mayor prominencia.

```python
wordcloud = WordCloud(background_color='white', width=1500, height=1000).generate('a b a c b c d a a e a b')
plt.figure(figsize=(5,5))
plt.imshow(wordcloud)
plt.axis('off')
plt.show()
```

```python
dictionary = {'a':5, 'b':3, 'c':2, 'd':1, 'e':1}
wordcloud = WordCloud(background_color='black',
                      width=1500,
                      height=1000
                      ).generate_from_frequencies(dictionary)

plt.figure(figsize=(5,5))
plt.imshow(wordcloud)
plt.axis('off')
plt.show()
```
