# Codificación visual

¿Con qué precisión podemos distinguir la diferencia entre los elementos codificados?

```python
xy1, xytext1 = (0,0), (1,0)
xy2, xytext2 = (0,1), (2,1)
```

¿Cuál es la diferencia entre la flecha de arriba y la de abajo?

```python
plt.annotate(text='', xy=xy1, xytext=xytext1, arrowprops=dict(arrowstyle='<->'))
plt.annotate(text='', xy=xy2, xytext=xytext2, arrowprops=dict(arrowstyle='<->'))
plt.axis('off')
plt.xlim([0, 2])
plt.show()
```

```python
x = [1,2]
y = [1,1]
c = 1000
s = [c,c*10]
```

¿Cuál es la diferencia entre el punto izquierdo y el derecho?

```python
plt.scatter(x,y,s=s,c='k')
plt.axis('off')
plt.xlim([0, 3])
plt.ylim([0, 2])
plt.show()
```

```python
def hist_and_lines():
    np.random.seed(0)
    fig, ax = plt.subplots(1, 2, figsize=(11, 4))
    ax[0].hist(np.random.randn(1000))
    for i in range(3):
        ax[1].plot(np.random.rand(10))
    ax[1].legend(['a', 'b', 'c'], loc='lower left')
```

```python
print(plt.style.available)
```

```python
hist_and_lines()
```

```python
with plt.style.context('fivethirtyeight'):
    hist_and_lines()
```

```python
with plt.style.context('classic'):
    hist_and_lines()
```

```python
with plt.style.context('ggplot'):
    hist_and_lines()
```

```python
with plt.style.context('dark_background'):
    hist_and_lines()
```

```python
with plt.style.context('grayscale'):
    hist_and_lines()
```
