Este dataset fue generado mediante simulaciones de 768 configuraciones de edificios residenciales en el software **Ecotect**, variando parámetros arquitectónicos clave. Fue publicado originalmente por Tsanas & Xifara (2012) en *Energy and Buildings*.

Puedes cargarlo directamente desde UCI con:

O usar el archivo `energy_efficiency.csv` provisto en el curso (mismo dataset con columnas en español).

### Variables Predictoras

| Variable | Nombre | Descripción |
|---|---|---|
| X1 | `CompacidadRelativa` | Relación volumen/superficie del edificio (0.62–0.98) |
| X2 | `AreaSuperficial` | Área total de la envolvente del edificio (m²) |
| X3 | `AreaMuros` | Área de muros exteriores (m²) |
| X4 | `AreaTecho` | Área del techo (m²) |
| X5 | `AlturaTotal` | Altura total del edificio — 2 valores: 3.5 m (1 piso) o 7 m (2 pisos) |
| X6 | `Orientacion` | Orientación cardinal codificada (2=Norte, 3=Este, 4=Sur, 5=Oeste) |
| X7 | `AreaAcristalamiento` | Proporción de superficie acristalada (0%, 10%, 25%, 40%) |
| X8 | `DistribAcristalamiento` | Distribución de ventanas en las fachadas (0=sin vidrio, 1–5=distintas distribuciones) |

### Variables Objetivo

| Variable | Tipo | Descripción |
|---|---|---|
| `CargaCalefaccion` | **Regresión** | Carga de calefacción requerida (kWh/m²) |
| `CargaRefrigeracion` | **Regresión** | Carga de refrigeración requerida (kWh/m²) |
| `ClaseEnergetica` | **Clasificación** | Categoría de eficiencia: `A` (alta eficiencia), `B` (media), `C` (baja eficiencia) |

> **Nota:** La variable `ClaseEnergetica` se deriva de `CargaCalefaccion` en terciles: clase A ≤ tercil 33%, clase C ≥ tercil 66%.  
> Algunas variables (`Orientacion`, `DistribAcristalamiento`) son categóricas ordinales y deberán tratarse apropiadamente antes del modelado.
